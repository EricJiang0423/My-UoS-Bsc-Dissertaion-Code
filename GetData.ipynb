Download the daily data of nasdaq stocks and crypto price datas

```python
import yfinance as yf
import pandas as pd
import os
import time
print("Start time: ", time.ctime())
# Delete the old data
os.system("rm -rf data")
os.system("mkdir data")
```

```python
# Create folders
stock_folder = "Data\Stock"
crypto_folder = "Data\Crypto"
```

```python

# make folders if they don't exist
os.makedirs(stock_folder, exist_ok=True)
os.makedirs(crypto_folder, exist_ok=True)

# Stock code list
stock_symbols = ["AAPL", "NVDA", "MSFT", "AMZN", "AVGO", "META", "TSLA", "COST", "NFLX", "GOOGL"]

# Cryptocurrency code list
crypto_symbols = ["BTC-USD", "ETH-USD", "XRP-USD", "USDT-USD", "BNB-USD", "DOGE-USD", "ADA-USD", "TRX-USD","XLM-USD"]

# Download and save data for each assets
for symbol in stock_symbols:
    data = yf.download(symbol, start="2015-01-01", end="2025-01-01", interval="1d", progress=True)
    data = data[['Open', 'High', 'Low', 'Close']]
    file_path = os.path.join(stock_folder, f"{symbol}.csv")
    data.to_csv(file_path)
    print(f"Stock Data {symbol} have been saved into {file_path}")

# Download and save data for each cryptocurrency
for symbol in crypto_symbols:
    data = yf.download(symbol, start="2015-01-01", end="2025-01-01", interval="1d", progress=True)
    data = data[['Open', 'High', 'Low', 'Close']]
    file_path = os.path.join(crypto_folder, f"{symbol}.csv")
    data.to_csv(file_path)
    print(f"Cryptocurrency Data {symbol} have been saved into {file_path}")

print("All data has been successfully downloaded and categorised for storage!")
```

```python
# Check download files for completeness, rate of missing data, and time span of data
def check_data_integrity(folder):
    for root, _, files in os.walk(folder):
        for file in files:
            file_path = os.path.join(root, file)
            data = pd.read_csv(file_path, index_col=0)
            missing_data = data.isnull().sum().sum()
            total_data = data.shape[0] * data.shape[1]
            missing_rate = missing_data / total_data
            start_date = data.index[2]
            end_date = data.index[-1]
            print(f"File {file} Missing data rate: {missing_rate:.2%}, Time span: {start_date} - {end_date}, Number of missing data: {missing_data}/{total_data}")

check_data_integrity(stock_folder)
check_data_integrity(crypto_folder)
```

```python
# Iterate over all the data and fill in the missing values using linear interpolation
def fill_missing_data(folder):
    for root, _, files in os.walk(folder):
        for file in files:
            file_path = os.path.join(root, file)
            data = pd.read_csv(file_path, index_col=0)
            # Processing data from the fourth row
            data.iloc[3:, data.columns.get_indexer(['Open', 'High', 'Low', 'Close'])] = data.iloc[3:, data.columns.get_indexer(['Open', 'High', 'Low', 'Close'])].astype(float).interpolate(method='linear')
            data.to_csv(file_path)
            print(f"File {file} missing values filled")

fill_missing_data(stock_folder)
fill_missing_data(crypto_folder)
```

```python
# Delete data prior to 2017-11-09
def remove_data(folder):
    for root, _, files in os.walk(folder):
        for file in files:
            file_path = os.path.join(root, file)
            data = pd.read_csv(file_path, index_col=0)
            data = data.loc['2017-11-09':]
            data.to_csv(file_path)
            print(f"File {file} Data removed")

remove_data(stock_folder)
remove_data(crypto_folder)
```

```python
# standardise the column names to: Open High Low Close and change the index column name to Date
def std_data(folder):
    for root, _, files in os.walk(folder):
        for file in files:
            file_path = os.path.join(root, file)
            data = pd.read_csv(file_path, index_col=0)
            data.index.name = 'Date'
            data.columns = ['Open', 'High', 'Low', 'Close']
            data.to_csv(file_path)
            print(f"File {file} Data Standardised")

std_data(stock_folder)
std_data(crypto_folder)
```

```python
# check the data span
def check_data_span(folder):
    for root, _, files in os.walk(folder):
        for file in files:
            file_path = os.path.join(root, file)
            data = pd.read_csv(file_path, index_col=0)
            start_date = data.index[0]
            end_date = data.index[-1]
            print(f"File {file} Time span: {start_date} - {end_date}")

check_data_span(stock_folder)
check_data_span(crypto_folder)
```
